```python
import os
import cv2
import matplotlib.pyplot as plt
```

```python
datadir='../data/raw/train'
classes=os.listdir(datadir)


fig,axis=plt.subplots(1,len(classes),figsize=(18,5))

for i,clas in enumerate(classes):    
    path=os.path.join(datadir,clas,'images')
    for img in os.listdir(path):
        a=cv2.imread(os.path.join(path,img))
        a=cv2.cvtColor(a,cv2.COLOR_BGR2RGB)
        axis[i].imshow(a) 
        axis[i,].set_title(clas)
        axis[i].axis('off')
        break
```

```python
datadir='../data/raw/train'
classes=os.listdir(datadir)
imagelens={}

for index,file in enumerate(classes):
    path=os.path.join(datadir,file,'images')
    files1=os.listdir(path)
    imagelens[classes[index]]=len(files1)
    
print(imagelens)
```

```python
for clas in classes:
    path=os.path.join(datadir,clas,'images')
    for img in os.listdir(path):
        a=cv2.imread(os.path.join(path,img))
        if a is None:
            print(f'cant read {img} in {clas}')
            os.remove(os.path.join(path,img))
        else:
            continue
```

```python
from collections import Counter

dimensions=[]
formats=[]

def count_elements(array):
    return Counter(array)

def imageshape(img):
    a=cv2.imread(img)
    return a.shape

def imageformat(img):
    a=os.path.splitext(img)
    return a[1]


for clas in classes:
    path=os.path.join(datadir,clas,'images')
    for img in os.listdir(path):
            dimensions.append(imageshape(os.path.join(path,img)))
            formats.append(imageformat(os.path.join(path,img)))

print(count_elements(dimensions))
print(count_elements(formats))
```
